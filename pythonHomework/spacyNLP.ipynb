```python
import spacy
nlp = spacy.load("en_core_web_lg")
filepath = 'Alexander_Warrior_Jar-220.txt'
f = open(filepath, 'r', encoding='utf8').read()
spacyRead = nlp(f)
for token in spacyRead:
    print(token.text, "---->", token.pos_, ":::::", token.lemma_)
spacy.explain("DET")
def wordCollector(words):
    wordList = []
    count = 0
    for token in words:
        if token.pos_ == "VERB":
            count += 1
            # print(count, ": ", token.text, " lemma: ", token.lemma_, " pos: ", token.pos_)
            # don't forget the underscore after token.lemma_ , token.pos_, etc.!
            wordList.append(token.lemma_)
            # print(count, ": ", token, token.pos_)
    # print(count, ": ", adjectives)
    return wordList
myWords = wordCollector(spacyRead)
print(myWords)
from collections import Counter

word_freq = Counter(myWords)
# most_common() converts the Counter's dictionary to a tuple series and sorts it
ranked = word_freq.most_common()
# print(ranked)
topTen = word_freq.most_common()[:10]
# Hey DIGIT 210! This is called slicing a list! You're taking the first 10 or 20 items from the list
# you can also stick the 10 in most_common(10), to get the first ten entries. 

# Try uncommenting these others to see how they go:
print(topTen)
# lastTen = word_freq.most_common()[:-21:-1]
# This is slicing off the END of the list the bottom 20. 
# print(lastTen)
o = open("verbFreq.txt", "w")
for w, num in ranked:
    o.write(str(w) + ' :::: ' + str(num) + "\n")
o = open("verbFreq2.txt", "w")
# ebb: on the next line, notice how we enumerate over the whole tuple: 
# the first variable will give us a position number, commonly known as an "index" called i.
for i, t in enumerate(ranked):
    o.write(str(i) + str(t) + "\n")
```
